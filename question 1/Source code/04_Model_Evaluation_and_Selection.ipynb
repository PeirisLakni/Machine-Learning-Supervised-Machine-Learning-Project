```python
# ============================================================
# COMPLETE MODEL EVALUATION AND SELECTION
# Hotel Booking Cancellation Prediction
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

from sklearn.model_selection import train_test_split

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    balanced_accuracy_score,
    matthews_corrcoef,
    log_loss,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    PrecisionRecallDisplay
)

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)


# ============================================================
# 2. CREATE OUTPUT FOLDERS
# ============================================================

os.makedirs(
    "model_results",
    exist_ok=True
)

os.makedirs(
    "models",
    exist_ok=True
)

os.makedirs(
    "model_results/plots",
    exist_ok=True
)


# ============================================================
# 3. LOAD VALIDATION MODEL COMPARISON RESULTS
# ============================================================

validation_results_path = (
    "model_results/"
    "validation_model_comparison.csv"
)

validation_results = pd.read_csv(
    validation_results_path
)

print("Validation results loaded successfully.")
print("Shape:", validation_results.shape)

display(
    validation_results.head()
)


# ============================================================
# 4. CHECK REQUIRED VALIDATION COLUMNS
# ============================================================

required_validation_columns = [
    "Model",
    "Sampling Method",
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "ROC AUC"
]

missing_validation_columns = [
    column
    for column in required_validation_columns
    if column not in validation_results.columns
]

if missing_validation_columns:

    raise KeyError(
        "The validation results file is missing: "
        f"{missing_validation_columns}"
    )


# ============================================================
# 5. SORT MODEL RESULTS
# ============================================================

validation_results = (
    validation_results
    .sort_values(
        by=[
            "ROC AUC",
            "F1 Score",
            "Recall"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

validation_results["Experiment"] = (
    validation_results["Model"].astype(str)
    + " + "
    + validation_results[
        "Sampling Method"
    ].astype(str)
)

print("\nValidation model comparison:")

display(
    validation_results.round(4)
)


# ============================================================
# 6. IDENTIFY BEST VALIDATION MODEL
# ============================================================

best_validation_row = validation_results.iloc[0]

best_model_name = (
    best_validation_row["Model"]
)

best_sampling_method = (
    best_validation_row["Sampling Method"]
)

best_experiment = (
    f"{best_model_name} + "
    f"{best_sampling_method}"
)

print("\n" + "=" * 70)
print("BEST VALIDATION MODEL")
print("=" * 70)

print(
    "Model:",
    best_model_name
)

print(
    "Sampling method:",
    best_sampling_method
)

print(
    "Validation accuracy:",
    round(
        best_validation_row["Accuracy"],
        4
    )
)

print(
    "Validation precision:",
    round(
        best_validation_row["Precision"],
        4
    )
)

print(
    "Validation recall:",
    round(
        best_validation_row["Recall"],
        4
    )
)

print(
    "Validation F1-score:",
    round(
        best_validation_row["F1 Score"],
        4
    )
)

print(
    "Validation ROC-AUC:",
    round(
        best_validation_row["ROC AUC"],
        4
    )
)


# ============================================================
# 7. PLOT MODEL COMPARISON
# ============================================================

metrics_to_plot = [
    "ROC AUC",
    "F1 Score",
    "Recall",
    "Precision",
    "Accuracy"
]

for metric in metrics_to_plot:

    plot_data = (
        validation_results
        .sort_values(
            metric,
            ascending=False
        )
        .head(12)
    )

    plt.figure(
        figsize=(14, 6)
    )

    plt.bar(
        plot_data["Experiment"],
        plot_data[metric]
    )

    plt.title(
        f"Top Model Combinations Based on {metric}"
    )

    plt.xlabel(
        "Model and Sampling Method"
    )

    plt.ylabel(metric)

    plt.xticks(
        rotation=70,
        ha="right"
    )

    plt.ylim(
        0,
        1
    )

    plt.tight_layout()

    plot_name = (
        metric.lower()
        .replace(" ", "_")
        .replace("-", "_")
    )

    plt.savefig(
        f"model_results/plots/"
        f"validation_{plot_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 8. COMPARE SAMPLING METHODS
# ============================================================

sampling_comparison = (
    validation_results
    .groupby(
        "Sampling Method"
    )[
        [
            "Accuracy",
            "Precision",
            "Recall",
            "F1 Score",
            "ROC AUC"
        ]
    ]
    .agg(
        ["mean", "max"]
    )
)

print(
    "\nPerformance comparison "
    "by sampling method:"
)

display(
    sampling_comparison.round(4)
)

sampling_comparison.to_csv(
    "model_results/"
    "sampling_method_comparison.csv"
)


# ============================================================
# 9. LOAD SAVED FINAL MODEL
# ============================================================

model_path = (
    "models/"
    "best_hotel_cancellation_model.pkl"
)

best_model = joblib.load(
    model_path
)

print(
    "\nSaved best model loaded successfully."
)

print(
    "Model type:",
    type(best_model)
)


# ============================================================
# 10. LOAD TEST PREDICTIONS
# ============================================================

test_predictions_path = (
    "model_results/"
    "test_predictions.csv"
)

test_predictions = pd.read_csv(
    test_predictions_path
)

required_prediction_columns = [
    "actual_is_canceled",
    "predicted_is_canceled",
    "cancellation_probability"
]

missing_prediction_columns = [
    column
    for column in required_prediction_columns
    if column not in test_predictions.columns
]

if missing_prediction_columns:

    raise KeyError(
        "The test prediction file is missing: "
        f"{missing_prediction_columns}"
    )

print(
    "\nTest predictions loaded successfully."
)

print(
    "Number of test observations:",
    len(test_predictions)
)

display(
    test_predictions[
        required_prediction_columns
    ].head()
)


# ============================================================
# 11. PREPARE ACTUAL VALUES AND PROBABILITIES
# ============================================================

y_actual = (
    test_predictions[
        "actual_is_canceled"
    ]
    .astype(int)
)

y_default_prediction = (
    test_predictions[
        "predicted_is_canceled"
    ]
    .astype(int)
)

y_probability = (
    test_predictions[
        "cancellation_probability"
    ]
    .astype(float)
    .clip(0.000001, 0.999999)
)


# ============================================================
# 12. SPLIT PREDICTIONS FOR THRESHOLD SELECTION
# ============================================================

# Threshold selection should not be performed on the same
# observations used for the final evaluation.
#
# Therefore, the saved test predictions are divided into:
# 50% threshold-selection set
# 50% final holdout evaluation set.

threshold_indices, holdout_indices = train_test_split(
    np.arange(
        len(test_predictions)
    ),
    test_size=0.50,
    random_state=42,
    stratify=y_actual
)

y_threshold_actual = (
    y_actual.iloc[
        threshold_indices
    ]
    .reset_index(drop=True)
)

y_threshold_probability = (
    y_probability.iloc[
        threshold_indices
    ]
    .reset_index(drop=True)
)

y_holdout_actual = (
    y_actual.iloc[
        holdout_indices
    ]
    .reset_index(drop=True)
)

y_holdout_probability = (
    y_probability.iloc[
        holdout_indices
    ]
    .reset_index(drop=True)
)

print(
    "\nThreshold-selection observations:",
    len(y_threshold_actual)
)

print(
    "Final holdout observations:",
    len(y_holdout_actual)
)


# ============================================================
# 13. METRIC CALCULATION FUNCTION
# ============================================================

def calculate_classification_metrics(
    y_true,
    y_pred,
    y_prob,
    threshold
):
    """
    Calculate detailed binary-classification metrics.
    """

    matrix = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )

    tn, fp, fn, tp = (
        matrix.ravel()
    )

    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0
    )

    negative_predictive_value = (
        tn / (tn + fn)
        if (tn + fn) > 0
        else 0
    )

    false_positive_rate = (
        fp / (fp + tn)
        if (fp + tn) > 0
        else 0
    )

    false_negative_rate = (
        fn / (fn + tp)
        if (fn + tp) > 0
        else 0
    )

    return {
        "Threshold": threshold,

        "Accuracy": accuracy_score(
            y_true,
            y_pred
        ),

        "Precision": precision_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "Recall": recall_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "Specificity": specificity,

        "Negative Predictive Value":
            negative_predictive_value,

        "F1 Score": f1_score(
            y_true,
            y_pred,
            zero_division=0
        ),

        "Balanced Accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred
            ),

        "ROC AUC": roc_auc_score(
            y_true,
            y_prob
        ),

        "PR AUC":
            average_precision_score(
                y_true,
                y_prob
            ),

        "Matthews Correlation Coefficient":
            matthews_corrcoef(
                y_true,
                y_pred
            ),

        "Log Loss": log_loss(
            y_true,
            y_prob,
            labels=[0, 1]
        ),

        "True Negatives": tn,
        "False Positives": fp,
        "False Negatives": fn,
        "True Positives": tp,

        "False Positive Rate":
            false_positive_rate,

        "False Negative Rate":
            false_negative_rate
    }


# ============================================================
# 14. THRESHOLD ANALYSIS
# ============================================================

threshold_values = np.arange(
    0.10,
    0.91,
    0.01
)

threshold_results = []

for threshold in threshold_values:

    threshold_prediction = (
        y_threshold_probability
        >= threshold
    ).astype(int)

    threshold_metric = (
        calculate_classification_metrics(
            y_true=y_threshold_actual,
            y_pred=threshold_prediction,
            y_prob=y_threshold_probability,
            threshold=threshold
        )
    )

    threshold_results.append(
        threshold_metric
    )

threshold_results_df = pd.DataFrame(
    threshold_results
)

print(
    "\nThreshold analysis completed."
)

display(
    threshold_results_df[
        [
            "Threshold",
            "Accuracy",
            "Precision",
            "Recall",
            "Specificity",
            "F1 Score",
            "Balanced Accuracy",
            "Matthews Correlation Coefficient"
        ]
    ].round(4)
)


# ============================================================
# 15. SELECT BEST THRESHOLD
# ============================================================

# Primary criterion: Highest F1-score
# Secondary criterion: Highest recall
# Third criterion: Highest balanced accuracy

threshold_results_sorted = (
    threshold_results_df
    .sort_values(
        by=[
            "F1 Score",
            "Recall",
            "Balanced Accuracy"
        ],
        ascending=False
    )
    .reset_index(drop=True)
)

best_threshold_row = (
    threshold_results_sorted
    .iloc[0]
)

best_threshold = float(
    best_threshold_row[
        "Threshold"
    ]
)

print("\n" + "=" * 70)
print("SELECTED DECISION THRESHOLD")
print("=" * 70)

print(
    "Selected threshold:",
    round(
        best_threshold,
        2
    )
)

print(
    "Threshold-selection F1-score:",
    round(
        best_threshold_row[
            "F1 Score"
        ],
        4
    )
)

print(
    "Threshold-selection precision:",
    round(
        best_threshold_row[
            "Precision"
        ],
        4
    )
)

print(
    "Threshold-selection recall:",
    round(
        best_threshold_row[
            "Recall"
        ],
        4
    )
)

print(
    "Threshold-selection balanced accuracy:",
    round(
        best_threshold_row[
            "Balanced Accuracy"
        ],
        4
    )
)


# ============================================================
# 16. PLOT THRESHOLD COMPARISON
# ============================================================

threshold_plot_metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score",
    "Balanced Accuracy"
]

for metric in threshold_plot_metrics:

    plt.figure(
        figsize=(9, 5)
    )

    plt.plot(
        threshold_results_df[
            "Threshold"
        ],
        threshold_results_df[
            metric
        ]
    )

    plt.axvline(
        x=best_threshold,
        linestyle="--",
        label=(
            f"Selected threshold = "
            f"{best_threshold:.2f}"
        )
    )

    plt.title(
        f"{metric} Across Decision Thresholds"
    )

    plt.xlabel(
        "Decision Threshold"
    )

    plt.ylabel(metric)

    plt.ylim(
        0,
        1
    )

    plt.legend()

    plt.grid(
        alpha=0.3
    )

    plt.tight_layout()

    plot_name = (
        metric.lower()
        .replace(" ", "_")
    )

    plt.savefig(
        f"model_results/plots/"
        f"threshold_{plot_name}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 17. FINAL HOLDOUT PREDICTION
# ============================================================

y_holdout_prediction = (
    y_holdout_probability
    >= best_threshold
).astype(int)


# ============================================================
# 18. FINAL HOLDOUT METRICS
# ============================================================

final_metrics = (
    calculate_classification_metrics(
        y_true=y_holdout_actual,
        y_pred=y_holdout_prediction,
        y_prob=y_holdout_probability,
        threshold=best_threshold
    )
)

final_metrics_df = pd.DataFrame(
    [final_metrics]
)

print("\n" + "=" * 70)
print("FINAL HOLDOUT EVALUATION")
print("=" * 70)

display(
    final_metrics_df[
        [
            "Threshold",
            "Accuracy",
            "Precision",
            "Recall",
            "Specificity",
            "F1 Score",
            "Balanced Accuracy",
            "ROC AUC",
            "PR AUC",
            "Matthews Correlation Coefficient",
            "Log Loss"
        ]
    ].round(4)
)


# ============================================================
# 19. FINAL CLASSIFICATION REPORT
# ============================================================

print(
    "\nClassification report:"
)

print(
    classification_report(
        y_holdout_actual,
        y_holdout_prediction,
        target_names=[
            "Not Cancelled",
            "Cancelled"
        ],
        zero_division=0
    )
)


# ============================================================
# 20. FINAL CONFUSION MATRIX
# ============================================================

final_confusion_matrix = (
    confusion_matrix(
        y_holdout_actual,
        y_holdout_prediction,
        labels=[0, 1]
    )
)

print(
    "\nFinal confusion matrix:"
)

print(
    final_confusion_matrix
)

final_tn, final_fp, final_fn, final_tp = (
    final_confusion_matrix.ravel()
)

print(
    "\nTrue negatives:",
    final_tn
)

print(
    "False positives:",
    final_fp
)

print(
    "False negatives:",
    final_fn
)

print(
    "True positives:",
    final_tp
)

ConfusionMatrixDisplay(
    confusion_matrix=
        final_confusion_matrix,

    display_labels=[
        "Not Cancelled",
        "Cancelled"
    ]
).plot()

plt.title(
    "Final Model Confusion Matrix\n"
    f"Threshold = {best_threshold:.2f}"
)

plt.tight_layout()

plt.savefig(
    "model_results/plots/"
    "final_confusion_matrix.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 21. ROC CURVE
# ============================================================

RocCurveDisplay.from_predictions(
    y_holdout_actual,
    y_holdout_probability,
    name=best_experiment
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--"
)

plt.title(
    "ROC Curve for Selected Model"
)

plt.tight_layout()

plt.savefig(
    "model_results/plots/"
    "final_roc_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 22. PRECISION-RECALL CURVE
# ============================================================

PrecisionRecallDisplay.from_predictions(
    y_holdout_actual,
    y_holdout_probability,
    name=best_experiment
)

baseline_precision = (
    y_holdout_actual.mean()
)

plt.axhline(
    y=baseline_precision,
    linestyle="--",
    label=(
        "Baseline cancellation rate"
    )
)

plt.title(
    "Precision-Recall Curve "
    "for Selected Model"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "model_results/plots/"
    "final_precision_recall_curve.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 23. PREDICTED PROBABILITY DISTRIBUTION
# ============================================================

plt.figure(
    figsize=(10, 6)
)

plt.hist(
    y_holdout_probability[
        y_holdout_actual == 0
    ],
    bins=30,
    alpha=0.6,
    label="Not Cancelled"
)

plt.hist(
    y_holdout_probability[
        y_holdout_actual == 1
    ],
    bins=30,
    alpha=0.6,
    label="Cancelled"
)

plt.axvline(
    x=best_threshold,
    linestyle="--",
    label=(
        f"Threshold = "
        f"{best_threshold:.2f}"
    )
)

plt.title(
    "Distribution of Predicted "
    "Cancellation Probabilities"
)

plt.xlabel(
    "Predicted Cancellation Probability"
)

plt.ylabel(
    "Number of Bookings"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    "model_results/plots/"
    "probability_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 24. ERROR ANALYSIS
# ============================================================

holdout_results = (
    test_predictions
    .iloc[
        holdout_indices
    ]
    .copy()
    .reset_index(drop=True)
)

holdout_results[
    "final_prediction"
] = y_holdout_prediction.values

holdout_results[
    "selected_threshold"
] = best_threshold

holdout_results[
    "prediction_status"
] = np.select(
    [
        (
            (
                holdout_results[
                    "actual_is_canceled"
                ] == 0
            )
            &
            (
                holdout_results[
                    "final_prediction"
                ] == 0
            )
        ),

        (
            (
                holdout_results[
                    "actual_is_canceled"
                ] == 0
            )
            &
            (
                holdout_results[
                    "final_prediction"
                ] == 1
            )
        ),

        (
            (
                holdout_results[
                    "actual_is_canceled"
                ] == 1
            )
            &
            (
                holdout_results[
                    "final_prediction"
                ] == 0
            )
        ),

        (
            (
                holdout_results[
                    "actual_is_canceled"
                ] == 1
            )
            &
            (
                holdout_results[
                    "final_prediction"
                ] == 1
            )
        )
    ],
    [
        "True Negative",
        "False Positive",
        "False Negative",
        "True Positive"
    ],
    default="Unknown"
)

error_summary = (
    holdout_results[
        "prediction_status"
    ]
    .value_counts()
    .rename_axis(
        "Prediction Status"
    )
    .reset_index(
        name="Count"
    )
)

error_summary[
    "Percentage"
] = (
    error_summary["Count"]
    / len(holdout_results)
    * 100
)

print(
    "\nPrediction error summary:"
)

display(
    error_summary.round(2)
)


# ============================================================
# 25. HIGH-CONFIDENCE ERRORS
# ============================================================

false_positives = (
    holdout_results[
        holdout_results[
            "prediction_status"
        ] == "False Positive"
    ]
    .sort_values(
        "cancellation_probability",
        ascending=False
    )
)

false_negatives = (
    holdout_results[
        holdout_results[
            "prediction_status"
        ] == "False Negative"
    ]
    .sort_values(
        "cancellation_probability",
        ascending=True
    )
)

print(
    "\nHighest-confidence false positives:"
)

display(
    false_positives.head(10)
)

print(
    "\nHighest-confidence false negatives:"
)

display(
    false_negatives.head(10)
)


# ============================================================
# 26. CREATE FINAL MODEL-SELECTION SUMMARY
# ============================================================

final_model_summary = pd.DataFrame(
    {
        "Selected Model": [
            best_model_name
        ],

        "Sampling Method": [
            best_sampling_method
        ],

        "Decision Threshold": [
            best_threshold
        ],

        "Accuracy": [
            final_metrics[
                "Accuracy"
            ]
        ],

        "Precision": [
            final_metrics[
                "Precision"
            ]
        ],

        "Recall": [
            final_metrics[
                "Recall"
            ]
        ],

        "Specificity": [
            final_metrics[
                "Specificity"
            ]
        ],

        "F1 Score": [
            final_metrics[
                "F1 Score"
            ]
        ],

        "ROC AUC": [
            final_metrics[
                "ROC AUC"
            ]
        ],

        "PR AUC": [
            final_metrics[
                "PR AUC"
            ]
        ],

        "Balanced Accuracy": [
            final_metrics[
                "Balanced Accuracy"
            ]
        ],

        "MCC": [
            final_metrics[
                "Matthews Correlation Coefficient"
            ]
        ]
    }
)

print(
    "\nFinal model-selection summary:"
)

display(
    final_model_summary.round(4)
)


# ============================================================
# 27. SAVE RESULTS
# ============================================================

validation_results.to_csv(
    "model_results/"
    "sorted_validation_model_comparison.csv",
    index=False
)

threshold_results_df.to_csv(
    "model_results/"
    "threshold_comparison.csv",
    index=False
)

final_metrics_df.to_csv(
    "model_results/"
    "final_model_evaluation.csv",
    index=False
)

final_model_summary.to_csv(
    "model_results/"
    "final_model_selection_summary.csv",
    index=False
)

holdout_results.to_csv(
    "model_results/"
    "final_holdout_predictions.csv",
    index=False
)

error_summary.to_csv(
    "model_results/"
    "prediction_error_summary.csv",
    index=False
)

false_positives.to_csv(
    "model_results/"
    "false_positive_predictions.csv",
    index=False
)

false_negatives.to_csv(
    "model_results/"
    "false_negative_predictions.csv",
    index=False
)


# ============================================================
# 28. SAVE FINAL MODEL AND THRESHOLD
# ============================================================

joblib.dump(
    best_model,
    "models/"
    "final_hotel_cancellation_model.pkl"
)

joblib.dump(
    best_threshold,
    "models/"
    "final_decision_threshold.pkl"
)

final_model_information = {
    "selected_model":
        best_model_name,

    "sampling_method":
        best_sampling_method,

    "selected_threshold":
        best_threshold,

    "selection_metric":
        "F1 Score",

    "secondary_metrics": [
        "Recall",
        "Balanced Accuracy"
    ],

    "test_accuracy":
        final_metrics["Accuracy"],

    "test_precision":
        final_metrics["Precision"],

    "test_recall":
        final_metrics["Recall"],

    "test_f1_score":
        final_metrics["F1 Score"],

    "test_roc_auc":
        final_metrics["ROC AUC"],

    "test_pr_auc":
        final_metrics["PR AUC"]
}

joblib.dump(
    final_model_information,
    "models/"
    "final_model_information.pkl"
)


# ============================================================
# 29. VERIFY SAVED MODEL AND THRESHOLD
# ============================================================

loaded_final_model = joblib.load(
    "models/"
    "final_hotel_cancellation_model.pkl"
)

loaded_threshold = joblib.load(
    "models/"
    "final_decision_threshold.pkl"
)

print(
    "\nSaved model type:",
    type(loaded_final_model)
)

print(
    "Saved threshold:",
    loaded_threshold
)


# ============================================================
# 30. FINAL OUTPUT SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("MODEL EVALUATION AND SELECTION COMPLETED")
print("=" * 70)

print(
    "Selected model:",
    best_model_name
)

print(
    "Selected sampling method:",
    best_sampling_method
)

print(
    "Selected decision threshold:",
    round(
        best_threshold,
        2
    )
)

print(
    "Final accuracy:",
    round(
        final_metrics[
            "Accuracy"
        ],
        4
    )
)

print(
    "Final precision:",
    round(
        final_metrics[
            "Precision"
        ],
        4
    )
)

print(
    "Final recall:",
    round(
        final_metrics[
            "Recall"
        ],
        4
    )
)

print(
    "Final F1-score:",
    round(
        final_metrics[
            "F1 Score"
        ],
        4
    )
)

print(
    "Final ROC-AUC:",
    round(
        final_metrics[
            "ROC AUC"
        ],
        4
    )
)

print(
    "Final PR-AUC:",
    round(
        final_metrics[
            "PR AUC"
        ],
        4
    )
)

print("\nFiles created:")

print(
    "1. model_results/"
    "sorted_validation_model_comparison.csv"
)

print(
    "2. model_results/"
    "sampling_method_comparison.csv"
)

print(
    "3. model_results/"
    "threshold_comparison.csv"
)

print(
    "4. model_results/"
    "final_model_evaluation.csv"
)

print(
    "5. model_results/"
    "final_model_selection_summary.csv"
)

print(
    "6. model_results/"
    "final_holdout_predictions.csv"
)

print(
    "7. model_results/"
    "prediction_error_summary.csv"
)

print(
    "8. model_results/"
    "false_positive_predictions.csv"
)

print(
    "9. model_results/"
    "false_negative_predictions.csv"
)

print(
    "10. models/"
    "final_hotel_cancellation_model.pkl"
)

print(
    "11. models/"
    "final_decision_threshold.pkl"
)

print(
    "12. models/"
    "final_model_information.pkl"
)

print(
    "13. Evaluation plots saved in "
    "model_results/plots/"
)
```

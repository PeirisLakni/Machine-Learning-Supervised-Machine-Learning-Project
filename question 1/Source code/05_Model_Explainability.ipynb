```python
# ============================================================
# LOAD FEATURE-ENGINEERED CLEANED DATASET
# ============================================================

import pandas as pd
from sklearn.model_selection import train_test_split

data_path = (r"C:\Users\ranpa\Desktop\MSc in DS\Machine Learning\Coursework\Question_1\processed_data\hotel_bookings_feature_engineered.csv"
)

df = pd.read_csv(data_path)

print("Dataset shape:", df.shape)
print(df.head())


# ============================================================
# DEFINE TARGET
# ============================================================

target_column = "is_canceled"

X = df.drop(
    columns=[target_column]
)

y = df[target_column].astype(int)


# ============================================================
# RECREATE THE SAME SPLIT USED DURING TRAINING
# ============================================================

X_train_validation, X_test, y_train_validation, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)

X_train, X_validation, y_train, y_validation = (
    train_test_split(
        X_train_validation,
        y_train_validation,
        test_size=0.25,
        random_state=42,
        stratify=y_train_validation
    )
)

X_test = X_test.reset_index(drop=True)
y_test = y_test.reset_index(drop=True)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
# ============================================================
# COMPLETE MODEL EXPLAINABILITY
# Hotel Booking Cancellation Prediction
# ============================================================

# If SHAP is not installed, run once:
# !pip install shap

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import os
import joblib
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

warnings.filterwarnings("ignore")

from sklearn.inspection import permutation_importance
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

try:
    import shap
    SHAP_AVAILABLE = True

except ImportError:
    SHAP_AVAILABLE = False
    print(
        "SHAP is not installed. Run: !pip install shap"
    )


# ============================================================
# 2. CREATE OUTPUT FOLDERS
# ============================================================

os.makedirs(
    "model_explainability",
    exist_ok=True
)

os.makedirs(
    "model_explainability/plots",
    exist_ok=True
)


# ============================================================
# 3. LOAD THE FINAL MODEL
# ============================================================

model_path = (
    "models/"
    "best_hotel_cancellation_model.pkl"
)

if not os.path.exists(model_path):

    raise FileNotFoundError(
        f"Model not found at: {model_path}\n"
        "Run the model-selection notebook first."
    )

final_pipeline = joblib.load(
    model_path
)

print("Final model loaded successfully.")
print("Pipeline type:", type(final_pipeline))


# ============================================================
# 4. CHECK REQUIRED PIPELINE COMPONENTS
# ============================================================

if "preprocessor" not in final_pipeline.named_steps:

    raise KeyError(
        "The pipeline does not contain a "
        "'preprocessor' step."
    )

if "classifier" not in final_pipeline.named_steps:

    raise KeyError(
        "The pipeline does not contain a "
        "'classifier' step."
    )

fitted_preprocessor = (
    final_pipeline.named_steps[
        "preprocessor"
    ]
)

fitted_classifier = (
    final_pipeline.named_steps[
        "classifier"
    ]
)

print(
    "Final classifier:",
    type(fitted_classifier).__name__
)

print(
    "Pipeline steps:",
    list(final_pipeline.named_steps.keys())
)


# ============================================================
# 5. CHECK TEST DATA
# ============================================================

required_variables = [
    "X_test",
    "y_test"
]

missing_variables = [
    variable
    for variable in required_variables
    if variable not in globals()
]

if missing_variables:

    raise NameError(
        "The following variables are missing: "
        f"{missing_variables}. "
        "Load or recreate the test dataset before "
        "running explainability."
    )

X_explain = X_test.copy()

if isinstance(y_test, pd.Series):

    y_explain = (
        y_test
        .copy()
        .reset_index(drop=True)
    )

else:

    y_explain = pd.Series(
        y_test
    ).reset_index(drop=True)

X_explain = X_explain.reset_index(
    drop=True
)

print(
    "\nTest observations:",
    X_explain.shape[0]
)

print(
    "Original input features:",
    X_explain.shape[1]
)


# ============================================================
# 6. OBTAIN TRANSFORMED FEATURE NAMES
# ============================================================

try:

    transformed_feature_names = (
        fitted_preprocessor
        .get_feature_names_out()
    )

except Exception:

    transformed_sample = (
        fitted_preprocessor
        .transform(
            X_explain.iloc[:1]
        )
    )

    transformed_feature_names = np.array(
        [
            f"Feature_{index}"
            for index in range(
                transformed_sample.shape[1]
            )
        ]
    )

# Remove transformer prefixes for readability
clean_feature_names = []

for feature_name in transformed_feature_names:

    clean_name = str(feature_name)

    clean_name = clean_name.replace(
        "numerical__",
        ""
    )

    clean_name = clean_name.replace(
        "categorical__",
        ""
    )

    clean_feature_names.append(
        clean_name
    )

clean_feature_names = np.array(
    clean_feature_names
)

print(
    "Transformed features:",
    len(clean_feature_names)
)


# ============================================================
# 7. TRANSFORM TEST DATA
# ============================================================

X_test_transformed = (
    fitted_preprocessor
    .transform(
        X_explain
    )
)

# Convert sparse matrix if necessary
if hasattr(
    X_test_transformed,
    "toarray"
):

    X_test_transformed = (
        X_test_transformed.toarray()
    )

X_test_transformed = np.asarray(
    X_test_transformed
)

X_test_transformed_df = pd.DataFrame(
    X_test_transformed,
    columns=clean_feature_names
)

print(
    "Transformed test shape:",
    X_test_transformed_df.shape
)


# ============================================================
# 8. ORIGINAL-FEATURE PERMUTATION IMPORTANCE
# ============================================================

# This evaluates importance using the complete pipeline.
# Results remain at the original feature level.

print(
    "\nCalculating permutation importance..."
)

permutation_result = permutation_importance(
    estimator=final_pipeline,
    X=X_explain,
    y=y_explain,
    scoring="roc_auc",
    n_repeats=5,
    random_state=42,
    n_jobs=-1
)

permutation_importance_df = pd.DataFrame(
    {
        "Feature": X_explain.columns,

        "Importance Mean":
            permutation_result.importances_mean,

        "Importance Standard Deviation":
            permutation_result.importances_std
    }
)

permutation_importance_df = (
    permutation_importance_df
    .sort_values(
        "Importance Mean",
        ascending=False
    )
    .reset_index(drop=True)
)

print(
    "\nOriginal-feature permutation importance:"
)

display(
    permutation_importance_df
    .head(20)
    .round(5)
)

permutation_importance_df.to_csv(
    "model_explainability/"
    "permutation_importance.csv",
    index=False
)


# ============================================================
# 9. PLOT PERMUTATION IMPORTANCE
# ============================================================

top_permutation_features = (
    permutation_importance_df
    .head(20)
    .sort_values(
        "Importance Mean",
        ascending=True
    )
)

plt.figure(
    figsize=(10, 8)
)

plt.barh(
    top_permutation_features["Feature"],
    top_permutation_features[
        "Importance Mean"
    ]
)

plt.xlabel(
    "Decrease in ROC-AUC After Permutation"
)

plt.ylabel(
    "Original Feature"
)

plt.title(
    "Top 20 Features Based on "
    "Permutation Importance"
)

plt.tight_layout()

plt.savefig(
    "model_explainability/plots/"
    "permutation_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# ============================================================
# 10. EXTRACT MODEL-BASED FEATURE IMPORTANCE
# ============================================================

model_importance_df = None

# ------------------------------------------------------------
# Tree-based models
# ------------------------------------------------------------

if hasattr(
    fitted_classifier,
    "feature_importances_"
):

    model_importance_values = (
        fitted_classifier
        .feature_importances_
    )

    model_importance_df = pd.DataFrame(
        {
            "Feature":
                clean_feature_names,

            "Importance":
                model_importance_values
        }
    )

    importance_type = (
        "Tree Feature Importance"
    )

# ------------------------------------------------------------
# Linear models
# ------------------------------------------------------------

elif hasattr(
    fitted_classifier,
    "coef_"
):

    coefficient_values = (
        fitted_classifier.coef_
    )

    if coefficient_values.ndim == 2:

        coefficient_values = (
            coefficient_values[0]
        )

    model_importance_df = pd.DataFrame(
        {
            "Feature":
                clean_feature_names,

            "Coefficient":
                coefficient_values,

            "Absolute Importance":
                np.abs(
                    coefficient_values
                )
        }
    )

    model_importance_df[
        "Direction"
    ] = np.where(
        model_importance_df[
            "Coefficient"
        ] > 0,
        "Increases cancellation probability",
        "Decreases cancellation probability"
    )

    importance_type = (
        "Logistic Regression Coefficients"
    )


# ============================================================
# 11. DISPLAY AND SAVE MODEL IMPORTANCE
# ============================================================

if model_importance_df is not None:

    if "Importance" in model_importance_df.columns:

        model_importance_df = (
            model_importance_df
            .sort_values(
                "Importance",
                ascending=False
            )
            .reset_index(drop=True)
        )

        plot_value_column = (
            "Importance"
        )

    else:

        model_importance_df = (
            model_importance_df
            .sort_values(
                "Absolute Importance",
                ascending=False
            )
            .reset_index(drop=True)
        )

        plot_value_column = (
            "Absolute Importance"
        )

    print(
        f"\n{importance_type}:"
    )

    display(
        model_importance_df
        .head(25)
        .round(5)
    )

    model_importance_df.to_csv(
        "model_explainability/"
        "model_feature_importance.csv",
        index=False
    )

    top_model_features = (
        model_importance_df
        .head(20)
        .sort_values(
            plot_value_column,
            ascending=True
        )
    )

    plt.figure(
        figsize=(10, 8)
    )

    plt.barh(
        top_model_features["Feature"],
        top_model_features[
            plot_value_column
        ]
    )

    plt.xlabel(
        plot_value_column
    )

    plt.ylabel(
        "Transformed Feature"
    )

    plt.title(
        f"Top 20 Features: {importance_type}"
    )

    plt.tight_layout()

    plt.savefig(
        "model_explainability/plots/"
        "model_feature_importance.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

else:

    print(
        "\nThe selected classifier does not provide "
        "native feature importance or coefficients."
    )


# ============================================================
# 12. LOGISTIC REGRESSION DIRECTION PLOT
# ============================================================

if hasattr(
    fitted_classifier,
    "coef_"
):

    coefficient_plot_df = (
        model_importance_df
        .head(20)
        .sort_values(
            "Coefficient",
            ascending=True
        )
    )

    plt.figure(
        figsize=(10, 8)
    )

    plt.barh(
        coefficient_plot_df[
            "Feature"
        ],
        coefficient_plot_df[
            "Coefficient"
        ]
    )

    plt.axvline(
        x=0,
        linestyle="--"
    )

    plt.xlabel(
        "Model Coefficient"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Features Increasing and Decreasing "
        "Cancellation Probability"
    )

    plt.tight_layout()

    plt.savefig(
        "model_explainability/plots/"
        "coefficient_direction.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 13. PREPARE SHAP SAMPLE
# ============================================================

if SHAP_AVAILABLE:

    # Use a manageable sample because SHAP can be expensive
    shap_sample_size = min(
        500,
        len(X_test_transformed_df)
    )

    X_shap = (
        X_test_transformed_df
        .sample(
            n=shap_sample_size,
            random_state=42
        )
        .reset_index(drop=True)
    )

    print(
        "\nSHAP sample size:",
        len(X_shap)
    )


# ============================================================
# 14. CREATE SHAP EXPLAINER
# ============================================================

if SHAP_AVAILABLE:

    shap_values_for_plot = None
    shap_explanation = None

    try:

        # ----------------------------------------------------
        # Tree-based models
        # ----------------------------------------------------

        if isinstance(
            fitted_classifier,
            (
                DecisionTreeClassifier,
                RandomForestClassifier
            )
        ) or hasattr(
            fitted_classifier,
            "feature_importances_"
        ):

            shap_explainer = shap.TreeExplainer(
                fitted_classifier
            )

            raw_shap_values = (
                shap_explainer.shap_values(
                    X_shap
                )
            )

            # Different SHAP versions return different formats
            if isinstance(
                raw_shap_values,
                list
            ):

                if len(raw_shap_values) == 2:

                    shap_values_for_plot = (
                        raw_shap_values[1]
                    )

                else:

                    shap_values_for_plot = (
                        raw_shap_values[0]
                    )

            else:

                raw_shap_values = np.asarray(
                    raw_shap_values
                )

                if raw_shap_values.ndim == 3:

                    shap_values_for_plot = (
                        raw_shap_values[:, :, 1]
                    )

                else:

                    shap_values_for_plot = (
                        raw_shap_values
                    )

        # ----------------------------------------------------
        # Logistic regression and other linear models
        # ----------------------------------------------------

        elif isinstance(
            fitted_classifier,
            LogisticRegression
        ) or hasattr(
            fitted_classifier,
            "coef_"
        ):

            background_size = min(
                200,
                len(X_shap)
            )

            X_background = shap.sample(
                X_shap,
                background_size,
                random_state=42
            )

            shap_explainer = (
                shap.LinearExplainer(
                    fitted_classifier,
                    X_background
                )
            )

            shap_explanation = (
                shap_explainer(
                    X_shap
                )
            )

            shap_values_for_plot = (
                shap_explanation.values
            )

        # ----------------------------------------------------
        # Generic fallback
        # ----------------------------------------------------

        else:

            background_size = min(
                100,
                len(X_shap)
            )

            X_background = shap.sample(
                X_shap,
                background_size,
                random_state=42
            )

            shap_explainer = shap.Explainer(
                fitted_classifier.predict_proba,
                X_background
            )

            shap_explanation = shap_explainer(
                X_shap
            )

            shap_values_for_plot = (
                shap_explanation.values
            )

            if (
                shap_values_for_plot.ndim == 3
            ):

                shap_values_for_plot = (
                    shap_values_for_plot[:, :, 1]
                )

        shap_values_for_plot = np.asarray(
            shap_values_for_plot
        )

        print(
            "SHAP values calculated successfully."
        )

        print(
            "SHAP values shape:",
            shap_values_for_plot.shape
        )

    except Exception as error:

        shap_values_for_plot = None

        print(
            "\nSHAP calculation failed:"
        )

        print(
            type(error).__name__,
            "-",
            error
        )


# ============================================================
# 15. SHAP SUMMARY PLOT
# ============================================================

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    shap.summary_plot(
        shap_values_for_plot,
        X_shap,
        feature_names=clean_feature_names,
        max_display=20,
        show=False
    )

    plt.title(
        "SHAP Summary Plot: "
        "Hotel Cancellation Prediction"
    )

    plt.tight_layout()

    plt.savefig(
        "model_explainability/plots/"
        "shap_summary_plot.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 16. SHAP GLOBAL BAR PLOT
# ============================================================

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    shap.summary_plot(
        shap_values_for_plot,
        X_shap,
        feature_names=clean_feature_names,
        plot_type="bar",
        max_display=20,
        show=False
    )

    plt.title(
        "Global SHAP Feature Importance"
    )

    plt.tight_layout()

    plt.savefig(
        "model_explainability/plots/"
        "shap_global_importance.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 17. SAVE SHAP IMPORTANCE TABLE
# ============================================================

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    mean_absolute_shap = np.mean(
        np.abs(
            shap_values_for_plot
        ),
        axis=0
    )

    shap_importance_df = pd.DataFrame(
        {
            "Feature":
                clean_feature_names,

            "Mean Absolute SHAP Value":
                mean_absolute_shap
        }
    )

    shap_importance_df = (
        shap_importance_df
        .sort_values(
            "Mean Absolute SHAP Value",
            ascending=False
        )
        .reset_index(drop=True)
    )

    print(
        "\nGlobal SHAP importance:"
    )

    display(
        shap_importance_df
        .head(25)
        .round(5)
    )

    shap_importance_df.to_csv(
        "model_explainability/"
        "shap_feature_importance.csv",
        index=False
    )


# ============================================================
# 18. SHAP DEPENDENCE PLOTS
# ============================================================

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    top_shap_features = (
        shap_importance_df[
            "Feature"
        ]
        .head(5)
        .tolist()
    )

    for feature in top_shap_features:

        try:

            shap.dependence_plot(
                feature,
                shap_values_for_plot,
                X_shap,
                feature_names=
                    clean_feature_names,
                show=False
            )

            safe_feature_name = (
                str(feature)
                .replace("/", "_")
                .replace("\\", "_")
                .replace(" ", "_")
                .replace(":", "_")
            )

            plt.title(
                f"SHAP Dependence Plot: "
                f"{feature}"
            )

            plt.tight_layout()

            plt.savefig(
                "model_explainability/plots/"
                f"shap_dependence_"
                f"{safe_feature_name}.png",
                dpi=300,
                bbox_inches="tight"
            )

            plt.show()

        except Exception as error:

            print(
                "Dependence plot skipped for",
                feature,
                "-",
                error
            )


# ============================================================
# 19. EXPLAIN ONE INDIVIDUAL BOOKING
# ============================================================

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    observation_index = 0

    individual_values = (
        shap_values_for_plot[
            observation_index
        ]
    )

    individual_explanation_df = pd.DataFrame(
        {
            "Feature":
                clean_feature_names,

            "Feature Value":
                X_shap
                .iloc[
                    observation_index
                ]
                .values,

            "SHAP Value":
                individual_values,

            "Absolute SHAP Value":
                np.abs(
                    individual_values
                )
        }
    )

    individual_explanation_df[
        "Effect"
    ] = np.where(
        individual_explanation_df[
            "SHAP Value"
        ] > 0,
        "Increases cancellation prediction",
        "Decreases cancellation prediction"
    )

    individual_explanation_df = (
        individual_explanation_df
        .sort_values(
            "Absolute SHAP Value",
            ascending=False
        )
        .reset_index(drop=True)
    )

    print(
        "\nTop factors influencing "
        "one individual prediction:"
    )

    display(
        individual_explanation_df
        .head(15)
        .round(5)
    )

    individual_explanation_df.to_csv(
        "model_explainability/"
        "individual_booking_explanation.csv",
        index=False
    )


# ============================================================
# 20. PLOT INDIVIDUAL FEATURE CONTRIBUTIONS
# ============================================================

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    individual_plot_df = (
        individual_explanation_df
        .head(15)
        .sort_values(
            "SHAP Value",
            ascending=True
        )
    )

    plt.figure(
        figsize=(10, 8)
    )

    plt.barh(
        individual_plot_df[
            "Feature"
        ],
        individual_plot_df[
            "SHAP Value"
        ]
    )

    plt.axvline(
        x=0,
        linestyle="--"
    )

    plt.xlabel(
        "SHAP Contribution"
    )

    plt.ylabel(
        "Feature"
    )

    plt.title(
        "Individual Booking Explanation"
    )

    plt.tight_layout()

    plt.savefig(
        "model_explainability/plots/"
        "individual_booking_explanation.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()


# ============================================================
# 21. CREATE PREDICTION EXPLANATION TABLE
# ============================================================

y_predicted = final_pipeline.predict(
    X_explain
)

y_probability = (
    final_pipeline
    .predict_proba(
        X_explain
    )[:, 1]
)

prediction_explanation_df = pd.DataFrame(
    {
        "Actual Outcome":
            y_explain,

        "Predicted Outcome":
            y_predicted,

        "Cancellation Probability":
            y_probability
    }
)

prediction_explanation_df[
    "Actual Label"
] = prediction_explanation_df[
    "Actual Outcome"
].map(
    {
        0: "Not Cancelled",
        1: "Cancelled"
    }
)

prediction_explanation_df[
    "Predicted Label"
] = prediction_explanation_df[
    "Predicted Outcome"
].map(
    {
        0: "Not Cancelled",
        1: "Cancelled"
    }
)

prediction_explanation_df[
    "Prediction Correct"
] = (
    prediction_explanation_df[
        "Actual Outcome"
    ]
    ==
    prediction_explanation_df[
        "Predicted Outcome"
    ]
)

prediction_explanation_df[
    "Confidence"
] = np.where(
    prediction_explanation_df[
        "Predicted Outcome"
    ] == 1,

    prediction_explanation_df[
        "Cancellation Probability"
    ],

    1
    -
    prediction_explanation_df[
        "Cancellation Probability"
    ]
)

prediction_explanation_df.to_csv(
    "model_explainability/"
    "prediction_explanations.csv",
    index=False
)

print(
    "\nPrediction explanation sample:"
)

display(
    prediction_explanation_df
    .head(10)
    .round(4)
)


# ============================================================
# 22. HIGH-CONFIDENCE INCORRECT PREDICTIONS
# ============================================================

high_confidence_errors = (
    prediction_explanation_df[
        prediction_explanation_df[
            "Prediction Correct"
        ] == False
    ]
    .sort_values(
        "Confidence",
        ascending=False
    )
)

print(
    "\nHigh-confidence incorrect predictions:"
)

display(
    high_confidence_errors
    .head(20)
    .round(4)
)

high_confidence_errors.to_csv(
    "model_explainability/"
    "high_confidence_errors.csv",
    index=False
)


# ============================================================
# 23. CREATE EXPLAINABILITY SUMMARY
# ============================================================

summary_information = {
    "Model Type":
        type(
            fitted_classifier
        ).__name__,

    "Number of Original Features":
        X_explain.shape[1],

    "Number of Transformed Features":
        len(clean_feature_names),

    "Explainability Methods":
        (
            "Permutation importance, "
            "native model importance, SHAP"
            if SHAP_AVAILABLE
            else
            "Permutation importance and "
            "native model importance"
        ),

    "Top Permutation Feature":
        permutation_importance_df
        .iloc[0]["Feature"],

    "Top Permutation Importance":
        permutation_importance_df
        .iloc[0]["Importance Mean"]
}

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    summary_information[
        "Top SHAP Feature"
    ] = shap_importance_df.iloc[0][
        "Feature"
    ]

    summary_information[
        "Top Mean Absolute SHAP Value"
    ] = shap_importance_df.iloc[0][
        "Mean Absolute SHAP Value"
    ]

explainability_summary_df = pd.DataFrame(
    [
        summary_information
    ]
)

explainability_summary_df.to_csv(
    "model_explainability/"
    "explainability_summary.csv",
    index=False
)

print(
    "\nExplainability summary:"
)

display(
    explainability_summary_df.T
)


# ============================================================
# 24. VERIFY CREATED FILES
# ============================================================

print("\n" + "=" * 70)
print("MODEL EXPLAINABILITY COMPLETED")
print("=" * 70)

created_files = [
    "model_explainability/"
    "permutation_importance.csv",

    "model_explainability/"
    "prediction_explanations.csv",

    "model_explainability/"
    "high_confidence_errors.csv",

    "model_explainability/"
    "explainability_summary.csv",

    "model_explainability/plots/"
    "permutation_importance.png"
]

if model_importance_df is not None:

    created_files.append(
        "model_explainability/"
        "model_feature_importance.csv"
    )

if (
    SHAP_AVAILABLE
    and shap_values_for_plot is not None
):

    created_files.extend(
        [
            "model_explainability/"
            "shap_feature_importance.csv",

            "model_explainability/"
            "individual_booking_explanation.csv",

            "model_explainability/plots/"
            "shap_summary_plot.png",

            "model_explainability/plots/"
            "shap_global_importance.png",

            "model_explainability/plots/"
            "individual_booking_explanation.png"
        ]
    )

for file_path in created_files:

    if os.path.exists(file_path):

        print("Created:", file_path)

    else:

        print("Not created:", file_path)

print(
    "\nAll explainability outputs are saved in:"
)

print(
    os.path.abspath(
        "model_explainability"
    )
)
```

```python
# ============================================================
# MODEL COMPARISON: SARIMA vs LSTM vs XGBOOST
# ============================================================

import pandas as pd
import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error
)
```

```python
# ============================================================
# FUNCTION TO CALCULATE METRICS
# ============================================================

def calculate_metrics(actual, predicted):

    mae = mean_absolute_error(
        actual,
        predicted
    )


    rmse = np.sqrt(
        mean_squared_error(
            actual,
            predicted
        )
    )


    mape = np.mean(
        np.abs(
            (actual - predicted)
            /
            actual
        )
    ) * 100


    return mae, rmse, mape
```
